This notebook regenerates the manuscript figure package exclusively from the locked CSV/TSV outputs already present in the repository.

**Scientific controls**

- No model fitting, tuning, feature selection, candidate re-ranking, or DFT execution is performed.
- Main and supplementary figures are descriptive visualizations of locked outputs.
- The fixed-geometry Quantum ESPRESSO voltage is reported only from the accepted energy chain.
- Unconverged endpoint relaxations are not used as voltage evidence.


```python
from pathlib import Path
import sys

def find_repository_root(start: Path | None = None) -> Path:
    current = (start or Path.cwd()).resolve()
    for candidate in [current, *current.parents]:
        if (
            (candidate / "data" / "processed" / "notebook_01").is_dir()
            and (candidate / "results").is_dir()
            and (candidate / "dft_validation").is_dir()
        ):
            return candidate
    raise FileNotFoundError("Repository root not found.")

ROOT = find_repository_root()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

ROOT
```
